# SVM이란 무엇인가?
분류, 회귀, 특이치 탐지에 사용하는 머신러닝 모델이다. 학습할 때 마진이 큰 방향으로 학습한다.

예를 들어, 세모 마을과 네모 마을 사이에 경계선(도로)을 놓을 때

* 결정 경계: 클래스를 나누는 선 자체
* 마진: 결정 경계 양옆의 "여유 공간"
*   서포트 벡터: 마진 가장자리에 딱 걸쳐있는, 경계를 결정짓는 핵심 데이터 포인트들<br>

-> SVM은 "제일 널찍하고 안전하게 나누는 선" 찾기를 목표로 함.
서포트 벡터에 의해서만 결정된다는 것이 특징.  

# 5.1 선형 SVM 분류

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.svm import SVC
from sklearn import datasets

iris = datasets.load_iris(as_frame=True)
X=iris.data[["petal length (cm)","petal width (cm)"]].values
y=iris.target

setosa_or_versicolor = (y==0) | (y==1)
X=X[setosa_or_versicolor]
y=y[setosa_or_versicolor]

# SVM 분류 모델
svm_clf = SVC(kernel="linear", C=10**9)
svm_clf.fit(X, y)

# 일반 선형 분류기 (임의로 그은 나쁜 결정 경계 3개)
x0 = np.linspace(0, 5.5, 200)
pred_1 = 5 * x0 - 20
pred_2 = x0 - 1.8
pred_3 = 0.1 * x0 + 0.5

# SVM 결정 경계 + 마진 + 서포트 벡터를 그리는 함수
def plot_svc_decision_boundary(svm_clf, xmin, xmax):
    w = svm_clf.coef_[0]
    b = svm_clf.intercept_[0]

    # w0*x0 + w1*x1 + b = 0 → x1 = -w0/w1 * x0 - b/w1
    x0 = np.linspace(xmin, xmax, 200)
    decision_boundary = -w[0] / w[1] * x0 - b / w[1]

    margin = 1 / w[1]
    gutter_up = decision_boundary + margin
    gutter_down = decision_boundary - margin
    svs = svm_clf.support_vectors_

    plt.plot(x0, decision_boundary, "k-", linewidth=2, zorder=-2)
    plt.plot(x0, gutter_up, "k--", linewidth=2, zorder=-2)
    plt.plot(x0, gutter_down, "k--", linewidth=2, zorder=-2)
    plt.scatter(svs[:, 0], svs[:, 1], s=180, facecolors="#AAA", zorder=-1)

fig, axes = plt.subplots(ncols=2, figsize=(10, 2.7), sharey=True)

# 왼쪽: 일반 선형 분류
plt.sca(axes[0])
plt.plot(x0, pred_1, "g--", linewidth=2)
plt.plot(x0, pred_2, "m-", linewidth=2)
plt.plot(x0, pred_3, "r-", linewidth=2)
plt.plot(X[:, 0][y==1], X[:, 1][y==1], "bs", label="Iris versicolor")
plt.plot(X[:, 0][y==0], X[:, 1][y==0], "yo", label="Iris setosa")
plt.xlabel("Petal length")
plt.ylabel("Petal width")
plt.legend(loc="upper left")
plt.axis([0, 5.5, 0, 2])
plt.gca().set_aspect("equal")
plt.grid()

# 오른쪽: SVM 분류
plt.sca(axes[1])
plot_svc_decision_boundary(svm_clf, 0, 5.5)
plt.plot(X[:, 0][y==1], X[:, 1][y==1], "bs")
plt.plot(X[:, 0][y==0], X[:, 1][y==0], "yo")
plt.xlabel("Petal length")
plt.axis([0, 5.5, 0, 2])
plt.gca().set_aspect("equal")
plt.grid()

save_fig("large_margin_classification_plot")
plt.show()
```

왼쪽은 일반 선형분류로, 학습 데이터는 다 나누지만, 초록 점선처럼 데이터에 너무 가까운 선은 새 데이터에서 틀릴 가능성이 큼

오른쪽은 SVM 분류로, 실선이 결정 경계이며 점선 두 개가 마진의 양쪽 끝. 제일 널찍하고 안전하게 나누는 분류

스케일링: 특성마다 제각각인 값의 범위를 비슷한 수준으로 맞춰서, 모델이 단위나 크기에 휘둘리지 않고 각 특성을 공정하게 반영하도록 하는 전처리 과정

```python
from sklearn.preprocessing import StandardScaler

Xs = np.array([(1, 50), (5, 20), (3, 80), (5, 60)]).astype(np.float64)
ys = np.array([0, 0, 1, 1])
svm_clf = SVC(kernel="linear", C=100).fit(Xs, ys)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(Xs)
svm_clf_scaled = SVC(kernel="linear", C=100).fit(X_scaled, ys)

plt.figure(figsize=(9, 2.7))

# 왼쪽: 스케일링 안 한 경우
plt.subplot(121)
plt.plot(Xs[:, 0][ys==1], Xs[:, 1][ys==1], "bo")
plt.plot(Xs[:, 0][ys==0], Xs[:, 1][ys==0], "ms")
plot_svc_decision_boundary(svm_clf, 0, 6)
plt.xlabel("$x_0$")
plt.ylabel("$x_1$    ", rotation=0)
plt.title("Unscaled")
plt.axis([0, 6, 0, 90])
plt.grid()

# 오른쪽: 스케일링 한 경우
plt.subplot(122)
plt.plot(X_scaled[:, 0][ys==1], X_scaled[:, 1][ys==1], "bo")
plt.plot(X_scaled[:, 0][ys==0], X_scaled[:, 1][ys==0], "ms")
plot_svc_decision_boundary(svm_clf_scaled, -2, 2)
plt.xlabel("$x'_0$")
plt.ylabel("$x'_1$  ", rotation=0)
plt.title("Scaled")
plt.axis([-2, 2, -2, 2])
plt.grid()

save_fig("sensitivity_to_feature_scales_plot")
plt.show()
```

**왼쪽은 스케일링을 하지 않은 경우**<br> >> x₁의 값이 커서 SVM이 x₁만 보고 경계를 긋기 때문에 경계가 거의 수평이고 x₀는 무시됨<br>

 **오른쪽은 스케일링을 한 경우**<br> >> 두 특성이 공정하게 반영돼서 경계가 기울어지고 마진도 더 균형 있게 잡힘

 >*즉 SVM은 특성의 크기에 민감하기 때문에 스케일링이 필수*



SVM의 두 가지 분류방식: 마진 위반 허용 정도에 따른 분류


1.   하드 마진 분류
2.   소프트 마진 분류



```python
# 하드 마진 분류

X_outliers = np.array([[3.4, 1.3], [3.2, 0.8]])
y_outliers = np.array([0, 0])

Xo1 = np.concatenate([X, X_outliers[:1]], axis=0)
yo1 = np.concatenate([y, y_outliers[:1]], axis=0)

Xo2 = np.concatenate([X, X_outliers[1:]], axis=0)
yo2 = np.concatenate([y, y_outliers[1:]], axis=0)

svm_clf2 = SVC(kernel="linear", C=10**9)
svm_clf2.fit(Xo2, yo2)

fig, axes = plt.subplots(ncols=2, figsize=(10, 2.7), sharey=True)

plt.sca(axes[0])
plt.plot(Xo1[:, 0][yo1==1], Xo1[:, 1][yo1==1], "bs")
plt.plot(Xo1[:, 0][yo1==0], Xo1[:, 1][yo1==0], "yo")
plt.text(0.3, 1.0, "Impossible!", color="red", fontsize=18)
plt.xlabel("Petal length")
plt.ylabel("Petal width")
plt.annotate("Outlier",
             xy=(X_outliers[0][0], X_outliers[0][1]),
             xytext=(2.5, 1.7),
             ha="center",
             arrowprops=dict(facecolor="black", shrink=0.1))
plt.axis([0, 5.5, 0, 2])
plt.grid()

plt.sca(axes[1])
plt.plot(Xo2[:, 0][yo2==1], Xo2[:, 1][yo2==1], "bs")
plt.plot(Xo2[:, 0][yo2==0], Xo2[:, 1][yo2==0], "yo")
plot_svc_decision_boundary(svm_clf2, 0, 5.5)
plt.xlabel("Petal length")
plt.annotate("Outlier",
             xy=(X_outliers[1][0], X_outliers[1][1]),
             xytext=(3.2, 0.08),
             ha="center",
             arrowprops=dict(facecolor="black", shrink=0.1))
plt.axis([0, 5.5, 0, 2])
plt.grid()

save_fig("sensitivity_to_outliers_plot")
plt.show()
```

왼쪽: 선형으로 완벽히 나눌 수 없음. 하드마진으로는 해 자체를 찾지 못함<br>
오른쪽: 이상치 하나 때문에 마진이 좁아지고 경계 기울어짐(이상치에 민감)

소프트 마진 분류는 마진 오류를 허용하면서 결정 경계 도로의 폭을 최대로 함<br>


C: SVM 규제 파라미터
*   C를 줄이면 마진이 커지고 훈련에러 허용을 많이 함(Underfit 가능)
*   C를 늘리면 마진이 줄어들고 훈련에러 허용을 적게 함(Overfit 가능)



```python
from sklearn.svm import LinearSVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn import datasets
import numpy as np
import matplotlib.pyplot as plt

iris = datasets.load_iris(as_frame=True)
X = iris.data[["petal length (cm)", "petal width (cm)"]].values
y = (iris.target == 2)  # Iris virginica

scaler = StandardScaler()
svm_clf1 = LinearSVC(C=1, max_iter=10_000, dual=True, random_state=42)
svm_clf2 = LinearSVC(C=100, max_iter=10_000, dual=True, random_state=42)

scaled_svm_clf1 = make_pipeline(scaler, svm_clf1)
scaled_svm_clf2 = make_pipeline(scaler, svm_clf2)

scaled_svm_clf1.fit(X, y)
scaled_svm_clf2.fit(X, y)

b1 = svm_clf1.decision_function([-scaler.mean_ / scaler.scale_])
b2 = svm_clf2.decision_function([-scaler.mean_ / scaler.scale_])
w1 = svm_clf1.coef_[0] / scaler.scale_
w2 = svm_clf2.coef_[0] / scaler.scale_
svm_clf1.intercept_ = np.array([b1])
svm_clf2.intercept_ = np.array([b2])
svm_clf1.coef_ = np.array([w1])
svm_clf2.coef_ = np.array([w2])

t = y * 2 - 1
support_vectors_idx1 = (t * (X.dot(w1) + b1) < 1).ravel()
support_vectors_idx2 = (t * (X.dot(w2) + b2) < 1).ravel()
svm_clf1.support_vectors_ = X[support_vectors_idx1]
svm_clf2.support_vectors_ = X[support_vectors_idx2]

fig, axes = plt.subplots(ncols=2, figsize=(10, 3.2), sharey=True)

plt.sca(axes[0])
plt.plot(X[:, 0][y==1], X[:, 1][y==1], "g^", label="Iris virginica")
plt.plot(X[:, 0][y==0], X[:, 1][y==0], "bs", label="Iris versicolor")
plot_svc_decision_boundary(svm_clf1, 4, 5.9)
plt.xlabel("Petal length")
plt.ylabel("Petal width")
plt.legend(loc="upper left")
plt.title(f"$C = {svm_clf1.C}$")
plt.axis([4, 5.9, 0.8, 2.8])
plt.grid()

plt.sca(axes[1])
plt.plot(X[:, 0][y==1], X[:, 1][y==1], "g^")
plt.plot(X[:, 0][y==0], X[:, 1][y==0], "bs")
plot_svc_decision_boundary(svm_clf2, 4, 5.9)
plt.xlabel("Petal length")
plt.title(f"$C = {svm_clf2.C}$")
plt.axis([4, 5.9, 0.8, 2.8])
plt.grid()

save_fig("regularization_plot")
plt.show()
```

C=1일때 마진 위반을 더 허용하기 때문에 마진이 넓어지고, C=100일때 위반을 덜 허용하여 마진이 좁아짐.





> C는 마진 폭과 마진 위반 사이의 균형을 조절하는 하이퍼파라미터

# 5.2 비선형 SVM 분류
선형적으로 분류할 수 없는 **데이터셋을** 다룸


다항 특성 추가 -> 차원을 늘려 선형으로 분류



```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import PolynomialFeatures, StandardScaler
from sklearn.svm import LinearSVC

X_moons, y_moons = make_moons(n_samples=100, noise=0.15, random_state=42)

def plot_dataset(X, y, axes):
    plt.plot(X[:, 0][y==0], X[:, 1][y==0], "bs")
    plt.plot(X[:, 0][y==1], X[:, 1][y==1], "g^")
    plt.axis(axes)
    plt.grid(True)
    plt.xlabel("$x_1$")
    plt.ylabel("$x_2$", rotation=0)

def plot_predictions(clf, axes):
    x0s = np.linspace(axes[0], axes[1], 100)
    x1s = np.linspace(axes[2], axes[3], 100)
    x0, x1 = np.meshgrid(x0s, x1s)
    X_grid = np.c_[x0.ravel(), x1.ravel()]
    y_pred = clf.predict(X_grid).reshape(x0.shape)
    y_decision = clf.decision_function(X_grid).reshape(x0.shape)
    plt.contourf(x0, x1, y_pred, cmap=plt.cm.brg, alpha=0.2)
    plt.contourf(x0, x1, y_decision, cmap=plt.cm.brg, alpha=0.1)

poly2_svm_clf = make_pipeline(
    PolynomialFeatures(degree=2),
    StandardScaler(),
    LinearSVC(C=10, max_iter=10_000, random_state=42)
)
poly3_svm_clf = make_pipeline(
    PolynomialFeatures(degree=3),
    StandardScaler(),
    LinearSVC(C=10, max_iter=10_000, random_state=42)
)

poly2_svm_clf.fit(X_moons, y_moons)
poly3_svm_clf.fit(X_moons, y_moons)

fig, axes = plt.subplots(ncols=2, figsize=(10, 3.2), sharey=True)
axis_range = [-1.5, 2.7, -1, 1.5]

plt.sca(axes[0])
plot_predictions(poly2_svm_clf, axis_range)
plot_dataset(X_moons, y_moons, axis_range)
plt.title("degree=2")

plt.sca(axes[1])
plot_predictions(poly3_svm_clf, axis_range)
plot_dataset(X_moons, y_moons, axis_range)
plt.title("degree=3")
plt.ylabel("")

save_fig("polynomial_features_svm_plot")
plt.show()
```

**5.2.1. 다항식 커널**


*   다항특성을 추가하는 것은 모델을 느리게 만듦
*   커널 트릭은 특성을 추가하지 않고, 매우 높은 차수의 다항 특성을 추가한 것과 같은 결과를 얻게 해줌<br>
-> 샘플 사이의 내적을 통해: 고차원으로 변환한 것과 같은 결과를 훨씬 적은 계산으로 얻음



```python
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

# 3차 커널
poly_kernel_svm_clf = make_pipeline(
    StandardScaler(),
    SVC(kernel="poly", degree=3, coef0=1, C=5)
)
poly_kernel_svm_clf.fit(X_moons, y_moons)

# 10차 커널
poly100_kernel_svm_clf = make_pipeline(
    StandardScaler(),
    SVC(kernel="poly", degree=10, coef0=100, C=5)
)
poly100_kernel_svm_clf.fit(X_moons, y_moons)

fig, axes = plt.subplots(ncols=2, figsize=(10.5, 4), sharey=True)

plt.sca(axes[0])
plot_predictions(poly_kernel_svm_clf, [-1.5, 2.4, -1, 1.5])
plot_dataset(X_moons, y_moons, [-1.5, 2.4, -1, 1.5])
plt.title("degree=3, coef0=1, C=5")

plt.sca(axes[1])
plot_predictions(poly100_kernel_svm_clf, [-1.5, 2.4, -1, 1.5])
plot_dataset(X_moons, y_moons, [-1.5, 2.4, -1, 1.5])
plt.title("degree=10, coef0=100, C=5")
plt.ylabel("")

save_fig("moons_kernelized_polynomial_svc_plot")
plt.show()
```

다항 커널 모델이 과적합이면 차수를 줄이고, 과소적합이면 차수를 늘린다.

**5.2.2. 유사도 특성**
* 각 샘플이 특정 기준(랜드마크)과 얼마나 유사한지 계산하여 새로운 특성으로 추가
* 랜드마크는 랜덤하게 선택하거나 전략적으로 선택<br>
<br>
가우스 방사 기저 함수<br>
-랜드마크(m)에서 멀어지면 값이 급격히 작아지게 함<br>
-감마는 얼마나 빨리 0으로 수렴하게 할지 결정, 값이 크면 m에서 멀어질수록 급격히 0으로 수렴<br>
<br>
φ(x, ℓ) = exp(−γ‖x − ℓ‖²)

```python
from sklearn.svm import LinearSVR
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt

svm_reg1 = LinearSVR(epsilon=0.5, dual=True, random_state=42)
svm_reg2 = LinearSVR(epsilon=1.2, dual=True, random_state=42)
svm_reg1.fit(X, y)
svm_reg2.fit(X, y)

def find_support_vectors(svm_reg, X, y):
    y_pred = svm_reg.predict(X)
    off_margin = np.abs(y - y_pred) >= svm_reg.epsilon
    return np.argwhere(off_margin)

svm_reg1.support_ = find_support_vectors(svm_reg1, X, y)
svm_reg2.support_ = find_support_vectors(svm_reg2, X, y)

eps_x1 = 1
eps_y_pred = svm_reg1.predict([[eps_x1]])

def plot_svm_regression(svm_reg, X, y, axes):
    x1s = np.linspace(axes[0], axes[1], 100).reshape(100, 1)
    y_pred = svm_reg.predict(x1s)
    plt.plot(x1s, y_pred, "k-", linewidth=2, label=r"$\hat{y}$")
    plt.plot(x1s, y_pred + svm_reg.epsilon, "k--")
    plt.plot(x1s, y_pred - svm_reg.epsilon, "k--")
    plt.scatter(X[svm_reg.support_], y[svm_reg.support_], s=180,
                facecolors="#AAA")
    plt.plot(X, y, "bo")
    plt.xlabel("$x_1$")
    plt.legend(loc="upper left")
    plt.axis(axes)

fig, axes = plt.subplots(ncols=2, figsize=(9, 4), sharey=True)

plt.sca(axes[0])
plot_svm_regression(svm_reg1, X, y, [0, 2, 3, 11])
plt.title(f"epsilon={svm_reg1.epsilon}")
plt.ylabel("$y$", rotation=0)
plt.annotate(
    "", xy=(eps_x1, eps_y_pred), xycoords="data",
    xytext=(eps_x1, eps_y_pred - svm_reg1.epsilon),
    textcoords="data",
    arrowprops={"arrowstyle": "<->", "linewidth": 1.5}
)
plt.text(0.9, 5.1, r"$\epsilon$", fontsize=13)

plt.sca(axes[1])
plot_svm_regression(svm_reg2, X, y, [0, 2, 3, 11])
plt.title(f"epsilon={svm_reg2.epsilon}")

save_fig("svm_regression_plot")
plt.show()
```



*   마진 안의 데이터: 예츩이 충분히 정확하므로, 모델이 그 데이터에 대해 학습하거나 예측을 수정하지 않음
*   마진 밖의 데이터: 예측 값과 실제 값의 차이가 큰 데이터로, 모델이 학습하고 예측을 조정하는 데 중요한 역할



```python
from sklearn.svm import SVR
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
X = 2 * np.random.rand(50, 1) - 1
y = 0.2 + 0.1 * X[:, 0] + 0.5 * X[:, 0] ** 2 + np.random.randn(50) / 10

svm_poly_reg1 = make_pipeline(StandardScaler(),
                               SVR(kernel="poly", degree=2, C=0.01, epsilon=0.1))
svm_poly_reg2 = make_pipeline(StandardScaler(),
                               SVR(kernel="poly", degree=2, C=100, epsilon=0.1))
svm_poly_reg1.fit(X, y)
svm_poly_reg2.fit(X, y)

def plot_svm_regression(svm_reg, X, y, axes):
    x1s = np.linspace(axes[0], axes[1], 100).reshape(100, 1)
    y_pred = svm_reg.predict(x1s)
    epsilon = svm_reg[-1].epsilon if hasattr(svm_reg, "__getitem__") else svm_reg.epsilon
    plt.plot(x1s, y_pred, "k-", linewidth=2, label=r"$\hat{y}$")
    plt.plot(x1s, y_pred + epsilon, "k--")
    plt.plot(x1s, y_pred - epsilon, "k--")
    plt.plot(X, y, "bo")
    plt.xlabel("$x_1$")
    plt.legend(loc="upper left")
    plt.axis(axes)

fig, axes = plt.subplots(ncols=2, figsize=(9, 4), sharey=True)

plt.sca(axes[0])
plot_svm_regression(svm_poly_reg1, X, y, [-1, 1, 0, 1])
plt.title(f"degree=2, C={svm_poly_reg1[-1].C}, "
          f"epsilon={svm_poly_reg1[-1].epsilon}")
plt.ylabel("$y$", rotation=0)

plt.sca(axes[1])
plot_svm_regression(svm_poly_reg2, X, y, [-1, 1, 0, 1])
plt.title(f"degree=2, C={svm_poly_reg2[-1].C}, "
          f"epsilon={svm_poly_reg2[-1].epsilon}")

save_fig("svm_with_polynomial_kernel_plot")
plt.show()
```

*   결과가 0보다 크면 예측된 클래스, 0보다 작으면 음성 클래스
* 결정경계는 결정함수의 값이 0인 선
* 마진의 가장자리는 결정함수의 값이 +1, -1인 선<br>
<br>
결정 함수<br>
h(x) = wᵀx + b = w₁x₁ + ⋯ + wₙxₙ + b



결정함수의 기울기로 w와 마진 크기 확인<br>
마진의 경계를 +1, -1이라고 정의
* 왼쪽 그래프 가중치 = 1

    - w₁x₁ = -1 또는 +1 : x₁ = -1, +1
    - 마진의 크기는 2

* 오른쪽 그래프 가중치 = 0.5
    - w₁x₁ = -1 또는 +1 : x₁ = -2, +2
    - 마진의 크기는 4

b는 마진의 크기에 영향을 주지 않으므로 일단 무시        

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(ncols=2, figsize=(10, 4), sharey=True)

for ax, w1 in zip(axes, [1, 0.5]):
    x1 = np.linspace(-3, 3, 200)
    s = w1 * x1

    ax.axhline(y=0, color="gray", linewidth=1)
    ax.axhline(y=1, color="k", linestyle="--", linewidth=1)
    ax.axhline(y=-1, color="k", linestyle="--", linewidth=1)
    ax.plot(x1, s, "b-", linewidth=2, label=r"$s = w_1 x_1$")

    margin_x = 1 / w1
    ax.axvspan(-margin_x, margin_x, color="gray", alpha=0.2)
    ax.annotate(
        "", xy=(margin_x, -1.7), xytext=(-margin_x, -1.7),
        arrowprops=dict(arrowstyle="<->")
    )
    ax.text(0, -1.9, "Margin", ha="center")

    ax.set_title(f"$w_1 = {w1}$")
    ax.set_xlabel("$x_1$")
    ax.set_xlim(-3, 3)
    ax.set_ylim(-2, 2)
    ax.grid(True)
    ax.legend(loc="upper left")

axes[0].set_ylabel("$s$")
plt.tight_layout()
save_fig("small_w_large_margin_plot")
plt.show()
```

# 5.5 쌍대 문제
> SVM에서 주어진 원래의 최적화 문제를 변형하여 쌍대 문제로 바꾸면, 더 효율적이거나 해결하기 쉬운 방식으로 SVM을 훈련할 수 있음

$$\max_{\alpha} \sum_{i=1}^{n} \alpha_i - \frac{1}{2}\sum_{i=1}^{n}\sum_{j=1}^{n} \alpha_i \alpha_j y_i y_j (\mathbf{x}_i \cdot \mathbf{x}_j)$$

$$\sum_{i=1}^{n} \alpha_i y_i = 0, \quad 0 \le \alpha_i \le C$$<br>

w를 직접 찾는 문제 -> α라는 새로운 값을 찾는 더 쉬운 문제
