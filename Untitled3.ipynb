# 결정 트리
결정 트리란?
* 분류와 회귀 모두 가능한 지도학습 모델. 데이터를 스무고개처럼 질문으로 쪼개면서 예측함
* 스케일링이나 평균을 맞추는 전처리가 필요 없음 — 특성 하나씩 기준값으로 나누기 때문
* 화이트박스 모델 (SVM, 신경망은 블랙박스) — 왜 그렇게 예측했는지 사람이 해석 가능<br>
→ 랜덤 포레스트 등 앙상블 기법의 기본 구성 요소이기도 함

6.1 결정 트리 학습과 시각화<br>
* 사이킷런의 DecisionTreeClassifier로 학습
* export_graphviz() 등으로 트리 구조를 시각화 가능
노드 구성 요소
* gini — 지니 불순도 (해당 노드가 얼마나 순수한지)
* samples — 이 노드에 해당하는 훈련 샘플 수
* value — 각 클래스별 샘플 개수
* class — 이 노드가 예측하는 클래스 (value 중 다수 클래스)


6.2 예측하기
* 루트 노드에서 시작 → 각 노드의 질문(조건)에 따라 왼쪽/오른쪽으로 이동 → 리프 노드에 도달하면 그 클래스로 예측
* 리프 노드: 더 이상 질문(자식)이 없는 노드 = 최종 예측이 내려지는 지점<br>

지니 불순도 (Gini impurity)
Gᵢ = 1 − Σ pᵢ,ₖ²
* pᵢ,ₖ = i번째 노드에 있는 훈련 샘플 중 클래스 k에 속하는 샘플의 비율
* 0이면 완전히 순수 (한 클래스만 존재), 값이 클수록 여러 클래스가 섞여 있음


6.3 클래스 확률 추정
* 한 샘플이 특정 리프 노드에 도달하면, 그 노드 안 클래스별 비율을 확률로 추정
* 같은 리프 노드에 도달한 샘플들은 전부 같은 확률을 갖게 됨 (리프 노드 단위로만 구분되기 때문)

6.4 CART 훈련 알고리즘
* 사이킷런은 CART (Classification And Regression Tree) 알고리즘 사용
* 훈련 세트를 하나의 특성 k와 임곗값 tₖ로 두 서브셋으로 나눔
* 어떻게 (k, tₖ)를 고르나? → 가중치가 적용된 자식 노드의 불순도 합이 최소가 되는 (k, tₖ)를 탐색

J(k, tₖ) = (m_left/m)·G_left + (m_right/m)·G_right
* 이 과정을 자식 노드에서도 재귀적으로 반복 → 최대 깊이(max_depth)에 도달하거나 더 이상 불순도를 줄일 수 없을 때 중단<br>
→ 탐욕적 알고리즘(greedy algorithm) — 각 단계에서 최적을 찾지만, 전체적으로 최적의 트리라는 보장은 없음. 전역 최적 트리를 찾는 건 NP-완전 문제라 현실적으로 불가능


6.5 계산 복잡도
* 예측: 루트에서 리프까지 이동
→ 트리가 균형 잡혀 있다면 약 O(log₂(m)) 개의 노드를 거침 (특성 개수와 무관 → 매우 빠름)
* 훈련: 각 노드에서 모든 샘플의 모든 특성을 비교 → 약 O(n × m log₂(m))


6.6 지니 불순도 또는 엔트로피

Hᵢ = − Σ pᵢ,ₖ log₂(pᵢ,ₖ) (pᵢ,ₖ ≠ 0)
* 기본값은 지니, **criterion="entropy"**로 변경 가능
* 실제로는 큰 차이 없음. 지니가 계산이 조금 더 빠름
* 지니는 가장 빈도 높은 클래스를 한쪽으로 몰아주는 경향, 엔트로피는 조금 더 균형 잡힌 트리를 만드는 경향


6.7 규제 매개변수
* 결정 트리는 비파라미터 모델 (파라미터 개수가 훈련 전 정해지지 않음) → 제한 없이 키우면 훈련 데이터에 과적합되기 쉬움
* 규제(=자유도 제한)를 걸어야 과대적합을 방지

| 하이퍼파라미터 | 역할 |
|---|---|
| max_depth | 트리의 최대 깊이 제한 |
| min_samples_split | 노드를 분할하기 위해 필요한 최소 샘플 수 |
| min_samples_leaf | 리프 노드가 가져야 할 최소 샘플 수 |
| min_weight_fraction_leaf | min_samples_leaf와 유사 (가중치 비율로 지정) |
| max_leaf_nodes | 리프 노드의 최대 개수 |
| max_features | 각 노드에서 분할에 사용할 특성의 최대 개수 |

→ min_* 계열 값을 늘리거나 max_* 계열 값을 줄이면 → 모델에 대한 규제가 커짐 (과소적합 방향)


6.8 회귀
* DecisionTreeRegressor로 회귀도 가능
* 분류와의 차이: 각 노드가 클래스 대신 **예측값(그 노드에 속한 샘플들의 평균 y값)**을 가짐
* 분할 기준도 불순도(Gini) 대신 **MSE(평균제곱오차)**를 최소화하는 방향으로 분할
MSE_node = (1/m_node) Σ (ŷ_node − y⁽ⁱ⁾)²

→ 회귀에서도 규제 없이 키우면 과대적합 위험 동일 (min_samples_leaf 등으로 제어)


6.9 축 방향에 대한 민감성
* 결정 트리는 결정 경계가 항상 축에 수직 (예: x₁=3, x₂=1.5 같은 형태)
* 데이터를 회전시키면(예: 45도 회전) 원래는 단순했던 경계가 매우 복잡한 계단 모양이 되어버림
* 해결책: PCA 등으로 데이터의 방향을 먼저 정렬한 후 트리를 학습시키면 더 단순하고 일반화 잘 되는 트리를 얻을 수 있음

6.10 결정 트리의 분산 문제
* 결정 트리는 분산이 큼 (high variance) — 훈련 데이터를 조금만 바꾸거나, 랜덤 시드(random_state)만 바꿔도 완전히 다른 트리가 나올 수 있음
* 같은 데이터라도 사이킷런 알고리즘 자체가 각 노드에서 확률적으로 특성을 고르는 부분이 있어 매번 트리가 조금씩 달라짐<br>
→ 이 분산 문제를 해결하기 위해 등장하는 것이 랜덤 포레스트 — 여러 트리를 평균 내어 분산을 줄임
