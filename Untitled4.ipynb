# 7장. 앙상블 학습과 랜덤 포레스트
앙상블 학습이란?
* 여러 개의 예측기(분류기/회귀기)를 모아서 하나의 예측기보다 더 좋은 성능을 내는 기법
* 집단지성과 같은 원리 — 다수의 약한 학습기를 모으면 하나의 강한 학습기가 됨
* 앙상블에 참여하는 개별 예측기를 약한 학습기, 결합된 전체를 강한 학습기라고 부름<br>

→ 랜덤 포레스트, 배깅/페이스팅, 부스팅, 스태킹이 대표적인 앙상블 기법


7.1 투표 기반 분류기
* 서로 다른 종류의 분류기(로지스틱 회귀, SVM, 결정 트리 등) 여러 개를 훈련시키고, 예측을 투표로 모으는 방식

직접 투표
* 각 분류기의 예측 클래스 중 다수결로 최종 클래스 결정

간접 투표
* 각 분류기가 출력하는 클래스 확률을 평균해서, 확률이 가장 높은 클래스로 결정
* 확신이 높은 투표에 비중을 더 두기 때문에 직접 투표보다 성능이 좋은 경우가 많음
* 모든 분류기가 predict_proba()를 지원해야 사용 가능<br>

→ 앙상블이 잘 작동하려면 예측기들이 서로 최대한 독립적이어야 함 (같은 오차를 반복하지 않도록)


7.2 배깅과 페이스팅
* 같은 알고리즘을 사용하되, 훈련 세트의 서브셋을 무작위로 다르게 추출해서 여러 개의 예측기를 학습시키는 방식
| 구분 | 배깅 (Bagging) | 페이스팅 (Pasting) |
|---|---|---|
| 샘플링 방식 | 중복 허용 (복원 추출) | 중복 불허 (비복원 추출) |
| 이름 유래 | bootstrap aggregating | - |
* 모든 예측기가 훈련을 마치면, 예측은 모든 예측기의 예측을 모아 집계(분류: 최빈값, 회귀: 평균)
* 개별 예측기는 원본 훈련 세트로 학습한 것보다 편향이 크지만, 집계를 거치면 편향과 분산이 모두 감소

oob 평가 (out-of-bag)
* 배깅은 중복을 허용하므로, 한 예측기가 훈련에 사용하지 않은 샘플(평균 약 37%)이 생김
* 이 남는 샘플로 별도의 검증 세트 없이도 성능 평가 가능 → oob_score=True

랜덤 패치와 랜덤 서브스페이스
* 랜덤 패치 방식: 훈련 샘플과 특성을 모두 무작위로 샘플링
* 랜덤 서브스페이스 방식: 훈련 샘플은 전부 사용하고, 특성만 무작위로 샘플링<br>
→ 특성이 많은 고차원 데이터에서 유용, 편향은 늘지만 분산은 감소



7.3 랜덤 포레스트
* 일반적으로 배깅(또는 페이스팅)을 적용한 결정 트리의 앙상블
* 사이킷런에서는 BaggingClassifier + DecisionTreeClassifier 조합 대신, 최적화된 RandomForestClassifier를 바로 사용 가능
* 트리를 성장시킬 때 전체 특성 중 최적의 특성을 찾는 대신, 무작위로 선택한 특성 후보 중에서 최적을 탐색<br>
→ 트리들 간의 상관관계를 낮춰서 다양성을 높이고, 결과적으로 편향은 손해보지만 분산을 낮춰 더 좋은 모델을 만듦

엑스트라 트리 (Extremely Randomized Trees)
* 각 노드를 분할할 때 특성뿐 아니라 임곗값도 무작위로 설정
* 무작위성이 더 크기 때문에 편향은 늘고 분산은 더 줄어듦
* 최적의 임곗값을 찾는 과정이 없어 일반 랜덤 포레스트보다 훈련 속도가 빠름

특성 중요도
* 랜덤 포레스트는 각 특성이 불순도를 얼마나 감소시켰는지를 기준으로 특성 중요도를 자동으로 계산
* feature_importances_ 속성으로 확인 가능<br>
→ 특성 선택이나 데이터 이해에 유용하게 활용됨



7.4 부스팅 (Boosting)
* 약한 학습기 여러 개를 순차적으로 연결하여, 앞 모델의 오차를 다음 모델이 보완하도록 학습하는 방식
* 배깅/랜덤 포레스트와 달리 병렬이 아니라 순차적 — 이전 예측기의 결과가 다음 예측기 학습에 영향을 줌

에이다부스트 (AdaBoost)
* 이전 예측기가 과소적합했던(잘못 분류한) 훈련 샘플의 가중치를 높여서 다음 예측기가 그 샘플에 더 집중하도록 학습
* 모든 예측기 학습이 끝나면, 각 예측기의 성능(가중치)에 따라 가중 다수결 투표로 예측<br>
→ 순차적 학습이라 배깅처럼 병렬화 불가능 (확장성이 떨어짐)

그레이디언트 부스팅 (Gradient Boosting)
* 이전 예측기가 만든 **잔여 오차(residual error)**에 새로운 예측기를 학습시켜 나가는 방식
* 에이다부스트처럼 가중치를 수정하는 대신, 잔차 자체를 예측 대상으로 삼음
* learning_rate로 각 트리의 기여도를 조절 — 낮추면 트리를 더 많이 필요로 하지만 일반화 성능이 좋아지는 경향 (축소 규제, shrinkage)
* GradientBoostingRegressor, 대용량 데이터에는 히스토그램 기반의 HistGradientBoostingRegressor가 효율적<br>

→ 최적의 트리 개수는 조기 종료(early stopping) 기법으로 찾는 것이 일반적


7.5 스태킹 (Stacking)
* 앙상블의 예측을 단순 투표로 집계하는 대신, 집계하는 방법 자체를 학습시키는 방식
* 마지막에 각 예측기의 예측값을 입력으로 받아 최종 예측을 만드는 모델을 블렌더(blender) 또는 메타 학습기라고 부름
* 훈련 세트를 두 개의 서브셋으로 나눠서, 첫 번째 세트로 기반 예측기들을 학습하고 두 번째 세트에 대한 예측을 만든 뒤, 그 예측들을 새로운 특성으로 삼아 블렌더를 학습
* 사이킷런의 StackingClassifier / StackingRegressor로 구현 가능

→ 단순 투표보다 성능이 좋을 때가 많지만, 구조가 복잡해지고 훈련 시간이 오래 걸림
